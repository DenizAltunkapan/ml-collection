# k-Nearest Neighbors (kNN) on Iris Dataset

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
import numpy as np
```

## Exploring the Iris Dataset

The Iris dataset is a classic dataset in machine learning, commonly used for classification tasks. It contains measurements of iris flowers from **three species**:

- **Setosa**
- **Versicolor**
- **Virginica**

### Dataset Structure

- **Features (`data`)**: 4 numerical attributes for each sample:
  1. Sepal length (cm)
  2. Sepal width (cm)
  3. Petal length (cm)
  4. Petal width (cm)

- **Target (`target`)**: Integer labels representing the species:
  - 0 → Setosa
  - 1 → Versicolor
  - 2 → Virginica

- **Feature names**: `iris_dataset.feature_names`  
- **Target names**: `iris_dataset.target_names`  

```python
iris_dataset = load_iris()

# Inspect dataset keys, feature names, and target names
print("Keys of iris_dataset: \n{}".format(iris_dataset.keys()))
print("Feature names:", iris_dataset.feature_names)
print("Target names:", iris_dataset.target_names)

# Inspect last 10 samples and their labels
print("The following measurements:\n" + str(iris_dataset['data'][-10:]) +
      "\n\nCorrespond to the following labels:\n" + str(iris_dataset['target'][-10:]))
```

```python
# Split Data into Training and Test-Splits
# By default, train_test_split uses 25% of the data for testing and 75% for training
X_train, X_test, y_train, y_test = train_test_split(
    iris_dataset['data'], iris_dataset['target'],
    random_state=0
)

# Model with k=1 neighbor
knn1 = KNeighborsClassifier(n_neighbors=1)

# Model with k=3 neighbors
knn3 = KNeighborsClassifier(n_neighbors=3)

# 'fit' method trains the model on the training data (X_train, y_train)
# For kNN, this means storing the training data internally,
# since kNN is a lazy learner (no actual model parameters are learned)
knn1.fit(X_train, y_train)
knn3.fit(X_train, y_train)
```

```python
# Make Predictions based on test data
y_pred_knn1 = knn1.predict(X_test)
y_pred_knn3 = knn3.predict(X_test)

# Test Accurracy
print(f"Accuracy for KNN with 1 Neighbor: {np.mean(y_pred_knn1 == y_test)}")
print(f"Accuracy for KNN with 3 Neighbors: {np.mean(y_pred_knn3 == y_test)}")
```

```python
prediction = knn1.predict([[6, 3, 5, 2]])
print(prediction)
```

The accuracy for k=1 is the same as for k=3 because the Iris dataset is small and well-separated, so even a single neighbor is sufficient to classify most test points correctly.
