# k-Nearest Neighbors (kNN) Regression on a Synthetic 1-Feature Dataset

```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsRegressor
```

## Exploring the Synthetic Regression Dataset

This dataset is artificially generated using **scikit-learn's `make_regression`** function, commonly used for demonstrating regression methods.  

### Dataset Structure

- **Features (`X`)**: 1 numerical attribute per sample (can be extended to multiple features)
- **Target (`y`)**: Continuous numerical values generated using a linear relationship with optional random noise
- **Noise**: Random variation added to the target to simulate real-world deviations (`noise` parameter in `make_regression`)
- **Number of samples**: Defined by `n_samples` parameter (e.g., 100)
- **Number of features**: Defined by `n_features` parameter (e.g., 1 for simple visualization)
- **Note for kNN**: The kNN regressor predicts target values based on the **closest neighbors** rather than assuming a linear relationship, making it flexible for non-linear patterns.

```python
# Generate synthetic data: 1 feature, 100 samples, small noise
X, y = make_regression(n_samples=100, n_features=1, noise=10, random_state=0)

# Split Data into Training and Test Sets
# By default, train_test_split uses 25% of the data for testing and 75% for training
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=0)

# Define kNN Regressors
knn1 = KNeighborsRegressor(n_neighbors=1)
knn3 = KNeighborsRegressor(n_neighbors=3)

# Train the models
knn1.fit(X_train, y_train)
knn3.fit(X_train, y_train)
```

```python
# Make Predictions
y_pred_knn1 = knn1.predict(X_test)
y_pred_knn3 = knn3.predict(X_test)

# Visualize
plt.figure(figsize=(10,5))

# True values
plt.scatter(X_test, y_test, color='blue', label='True values')

# Predictions for k=1
plt.scatter(X_test, y_pred_knn1, color='red', label='kNN (k=1) predictions')

# Predictions for k=3
plt.scatter(X_test, y_pred_knn3, color='green', label='kNN (k=3) predictions')

plt.xlabel('Feature')
plt.ylabel('Target')
plt.title('k-Nearest Neighbors Regression')
plt.legend()
plt.show()
```

The kNN regression with 3 neighbors produces smoother predictions that better approximate the underlying trend, while 1 neighbor captures more local fluctuations but is noisier.
