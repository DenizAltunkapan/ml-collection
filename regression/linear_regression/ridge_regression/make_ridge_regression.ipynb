# Ridge Regression on a Synthetic 1-Feature Dataset

```python
from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split
from sklearn.linear_model import Ridge
import matplotlib.pyplot as plt
import numpy as np
```

## Exploring the Synthetic Regression Dataset

This dataset is artificially generated using **scikit-learn's `make_regression`** function, commonly used for demonstrating linear regression methods.  

### Dataset Structure

- **Features (`X`)**: 1 numerical attribute per sample (can be extended to multiple features)
- **Target (`y`)**: Continuous numerical values generated using a linear relationship with optional random noise
- **Noise**: Random variation added to the target to simulate real-world deviations (`noise` parameter in `make_regression`)
- **Number of samples**: Defined by `n_samples` parameter (e.g., 100)
- **Number of features**: Defined by `n_features` parameter (e.g., 1 for simple visualization)

```python
# Generate synthetic data: 1 feature, 100 samples, small noise
X, y = make_regression(n_samples=100, n_features=1, noise=10, random_state=0)

# Split Data into Training and Test Sets
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=0)
```

```python
# Ridge Regression with alpha = 10
ridge_model = Ridge(alpha=10)
ridge_model.fit(X_train, y_train)
```

```python
# Predictions
y_pred_ridge = ridge_model.predict(X_test)

# Visualize
plt.scatter(X_test, y_test, color='blue', label='True values')
plt.plot(X_test, y_pred_ridge, color='green', linestyle='--', label='Ridge Predicted')
plt.xlabel('Feature')
plt.ylabel('Target')
plt.title('Ridge Regression')
plt.legend()
plt.show()
```

```python
# Compare to Ordinary Leat Squares
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train, y_train)

y_pred_ls = model.predict(X_test)

# Visualize
plt.scatter(X_test, y_test, color='blue', label='True values')
plt.plot(X_test, y_pred_ridge, color='green', linestyle='--', label='Ridge Predicted')
plt.plot(X_test, y_pred_ls, color='red', linestyle='--', label='OSL Predicted')
plt.xlabel('Feature')
plt.ylabel('Target')
plt.title('Ridge Regression vs Ordinary Least Squares')
plt.legend()
plt.show()
```

Ridge Regression is effective when features are correlated or the dataset contains noise, as it reduces variance by shrinking coefficients.
Choosing an appropriate alpha balances bias and variance, leading to more stable predictions compared to ordinary least squares.
